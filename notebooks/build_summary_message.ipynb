```python
# Welcome to your new notebook
# Type here in the cell editor to add code!
from pyspark.sql.functions import col, desc
from datetime import datetime

# ----------------------------------------------------------------------
# Query: today's headlines with sentiment, top 10 by sentiment_compound
# ----------------------------------------------------------------------
df = spark.read.table("top_headlines_sentiment")
total_rows = df.count()

# Aggregate stats
sentiment_dist = (
    df.groupBy("sentiment_label")
      .count()
      .orderBy(desc("count"))
      .collect()
)
sentiment_summary = " | ".join([f"{r['sentiment_label']}: {r['count']}" for r in sentiment_dist])

# Top headlines table data — pick a reasonable subset
top_rows = (
    df.select("sentiment_label", "category", "source_name", "title", "sentiment_compound")
      .orderBy(col("sentiment_compound").asc())  # most negative first
      .limit(10)
      .collect()
)

# ----------------------------------------------------------------------
# Build HTML table — Teams renders this inline in the message body
# ----------------------------------------------------------------------
def color_for(label):
    return {"Positive": "#548235", "Neutral": "#7F7F7F", "Negative": "#C00000"}.get(label, "#000000")

rows_html = ""
for r in top_rows:
    title_safe = (r["title"] or "")[:80].replace("<", "&lt;").replace(">", "&gt;")
    src_safe   = (r["source_name"] or "").replace("<", "&lt;").replace(">", "&gt;")
    color      = color_for(r["sentiment_label"])
    rows_html += f"""
    <tr>
      <td style="padding:6px 10px;border:1px solid #ccc;color:{color};font-weight:bold;">{r['sentiment_label']}</td>
      <td style="padding:6px 10px;border:1px solid #ccc;">{r['category']}</td>
      <td style="padding:6px 10px;border:1px solid #ccc;">{src_safe}</td>
      <td style="padding:6px 10px;border:1px solid #ccc;">{title_safe}</td>
      <td style="padding:6px 10px;border:1px solid #ccc;text-align:right;">{r['sentiment_compound']:.2f}</td>
    </tr>"""

html_table = f"""
<h3>✅ News Pipeline ran successfully</h3>
<p><b>Total articles in Silver+Sentiment:</b> {total_rows}<br>
<b>Sentiment distribution:</b> {sentiment_summary}<br>
<b>Generated at:</b> {datetime.utcnow().strftime('%Y-%m-%d %H:%M UTC')}</p>

<p><b>Top 10 Headlines (most negative first):</b></p>
<table style="border-collapse:collapse;font-family:Arial,sans-serif;font-size:13px;">
  <thead>
    <tr style="background:#2E75B6;color:white;">
      <th style="padding:6px 10px;border:1px solid #ccc;text-align:left;">Sentiment</th>
      <th style="padding:6px 10px;border:1px solid #ccc;text-align:left;">Category</th>
      <th style="padding:6px 10px;border:1px solid #ccc;text-align:left;">Source</th>
      <th style="padding:6px 10px;border:1px solid #ccc;text-align:left;">Title</th>
      <th style="padding:6px 10px;border:1px solid #ccc;text-align:right;">Score</th>
    </tr>
  </thead>
  <tbody>{rows_html}</tbody>
</table>

<p style="margin-top:12px;color:#666;font-size:12px;">— Automated message from news_pipeline_orchestrator</p>
"""

# ----------------------------------------------------------------------
# Return the HTML so the pipeline can capture it
# ----------------------------------------------------------------------
print("Generated HTML message:")
print(html_table[:500] + "..." if len(html_table) > 500 else html_table)

# Critical: exit with the HTML payload so the pipeline can read it
mssparkutils.notebook.exit(html_table)
```
